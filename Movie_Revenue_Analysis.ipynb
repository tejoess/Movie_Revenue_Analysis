***Data gathering, cleaning and preparing***

```python
import pandas as pd

df=pd.read_csv('/content/tmdb_5000_movies.csv')
```

```python
df.head(10)
```

```python
df.shape
```

```python
df.describe()
```

```python
df.info()
```

```python
df.drop(columns=['homepage','tagline','original_title'])
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=['release_date'],inplace=True)
```

```python
df.fillna(df['runtime'].median(),inplace=True)
```

```python
df['overview'].fillna('No Overview available',inplace=True)
```

```python
df.isnull().sum()
```

```python
df[df['budget']==0].shape
```

```python
df[df['runtime']==0].shape
```

```python
df[df['popularity']==0].shape
```

```python
df[df['revenue']==0].shape
```

```python
df[df['vote_average']==0].shape
```

```python
df[df['vote_count']==0].shape
```

```python
df['budget']=df['budget'].replace(0,df['budget'].median())
df['runtime']=df['runtime'].replace(0,df['runtime'].median())
df['revenue'] = df['revenue'].replace(0, df['revenue'].median())
```

```python
df = df[df['vote_count'] > 0]
df['vote_average'] = df['vote_average'].replace(0, df['vote_average'].median())
```

```python
df.info()
```

```python
print(df.info())
print(df.describe())

```

***Data Visualization and Analysis***

```python
import matplotlib.pyplot as plt
import seaborn as sns
sample = df.head(50)
import seaborn as sns
```

```python
plt.scatter(df['budget'],df['revenue'])
plt.xlabel('budget')
plt.ylabel('revenue')
plt.title('Budget vs revenue')
plt.show()
```

There is only one movie with moderate budget and high revenue, most of the move has low budget and low revenue, some of them are have high budget but revenue is still below average and there are 3 movies with high budget and low revenue, that means
-High budget does not lead to high revenue

```python

topmovies=df.nlargest(10,'revenue')
plt.bar(topmovies['title'],topmovies['revenue'])
plt.xticks(rotation=90)
plt.title('Top 10 Highest-Grossing Movies')
plt.show()
```

```python
sns.relplot(data=topmovies,x=topmovies['budget'],y=topmovies['revenue'],kind='scatter',hue='title')
plt.title('Top 10 Highest-Grossing Movies')
```

There are marvel movies and animated and most of them are Scifi, so yess people love Scifi movies and have higher revenue



```python
corr_data=df[['budget', 'revenue']].corr()
sns.heatmap(corr_data)
plt.title('Heatmap of correlation between budget and revenue')
```

Budget and revenue are not strongly related, value is not near 0

```python
sns.scatterplot(data=df,x=df['revenue'],y=df['popularity'])
plt.title('Revenue vs Popularity')
```

Actually here most of the movies have less revenue and popularity, but i can see some movies has less revenue but more popularity whereas there is only one movie which has highest revenue but less popularity and other average movies

```python
sns.scatterplot(df,x=df['vote_average'],y=df['revenue'])
plt.title('Vote_average vs revenue')
```

According to this, high rated movies does not make high revenue at all, here mid range rating movie had highest revenue, RATING DOES NOT AFFECT REVENUE
